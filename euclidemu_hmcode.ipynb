```python
import matplotlib.pylab as plt
```

```python
import euclidemu2 as ee2

cosmo = {
    "As": 2.1e-9,
    "ns": 0.965,
    "Omb": 0.049,
    "Omm": 0.315,
    "h": 0.674,
    "mnu": 0.06,
    "w": -1.0,
    "wa": 0.0
}
ee3=ee2.PyEuclidEmulator()
z = [0.0, 1.0]

k, boost = ee3.get_boost(cosmo, z)

print(k[:5])
print(boost[0][:5])
```

```python
cosmo_par={'As':2.1e-09, 'ns':0.966, 'Omb':0.04,'Omm':0.3,'h':0.68,'mnu':0.15,'w':-1.0,'wa':0.0}
redshifts=[0,2,4,6,8,10]
k, b = ee3.get_boost(cosmo_par,redshifts)

for i in range(len(redshifts)):
    plt.loglog(k,b[i],label="z="+str(redshifts[i]))

plt.legend()
plt.show()
```
