```python
"""
Analytic Gaussian covariance matrix for the C_l vector from mcmc_model.py.
 
Uses the standard Knox (1995) formula for the variance of a band-power
estimate of an angular power spectrum:
 
    Var(C_l) = 2 / [(2l+1) * f_sky * Delta_l] * (C_l + N_l)^2
 
where:
  - C_l is your theoretical/fiducial convergence power spectrum
  - N_l = sigma_eps^2 / n_bar is the shape-noise power spectrum
  - f_sky is the fraction of sky covered by the survey
  - Delta_l is the width (in multipoles) of each ell-bin
 
This gives a DIAGONAL covariance matrix (bins assumed uncorrelated) --
the standard Gaussian approximation for a single (non-tomographic)
redshift bin with non-overlapping ell bins. It does NOT include the
non-Gaussian (trispectrum) contribution that becomes relevant at high
l / small scales, particularly once baryonic effects are involved --
if you need that, this is a starting point, not the final answer.
 
To regenerate for a different zmin, zmax, or k-range: this script
mutates the relevant globals inside the imported mcmc_model module
(ZMIN, ZMAX, REDSHIFT_NODES, Z_GRID, Z_GRID_R, k_custom) before calling
mcmc_model.model(...) to compute the fiducial C_l -- this works because
mcmc_model's functions look up these names in the module's own
namespace at call time, so overwriting mcmc_model.<name> changes what
they see, without needing to edit mcmc_model.py itself.
"""
```

```python
import numpy as np
import mcmc_model
import mcmc_model_no_baryons
import matplotlib.pylab as plt
```

```python

# ---------------------------------------------------------------------
# Euclid-like survey specs (adjust if yours differ)
# ---------------------------------------------------------------------
ASURVEY_DEG2 = 15000.0   # sky area, deg^2 (Euclid Table 4)
NGAL_ARCMIN2 = 30.0      # galaxy surface density, arcmin^-2
SIGMA_EPS = 0.30         # total intrinsic ellipticity dispersion
 
DEG2_TO_SR = (np.pi / 180.0) ** 2
ARCMIN2_PER_SR = (180.0 * 60.0 / np.pi) ** 2  # arcmin^2 per steradian
 
 
def _set_range(zmin, zmax, kmax, n_z_nodes=15, n_z_grid=300, n_z_grid_r=500, n_k=100):
    """Overwrite mcmc_model's z/k-range globals in place."""
    baryonic_pk_zmax = 1.5  # baccoemu baryonic emulator's hard limit
    if zmax > baryonic_pk_zmax:
        raise ValueError(
            f"zmax={zmax} exceeds the baryonic emulator's valid range "
            f"(z <= {baryonic_pk_zmax}). Either lower zmax, or use "
            f"model_dmo from mcmc_model_no_baryons.py instead, which "
            f"isn't limited the same way."
        )
 
    mcmc_model.ZMIN = zmin
    mcmc_model.ZMAX = zmax
    mcmc_model.REDSHIFT_NODES = np.linspace(zmin, zmax, n_z_nodes)
    mcmc_model.Z_GRID = np.linspace(zmin, zmax, n_z_grid)
    mcmc_model.Z_GRID_R = np.linspace(0.0, zmax, n_z_grid_r)
    mcmc_model.k_custom = np.logspace(-3.5, np.log10(kmax), num=n_k)
 
 
```

```python
def _set_range_dmo(zmin, zmax, kmax, n_z_nodes=15, n_z_grid=300, n_z_grid_r=500, n_k=100):
    """Overwrite mcmc_model's z/k-range globals in place."""
    baryonic_pk_zmax = 1.5  # baccoemu baryonic emulator's hard limit
    if zmax > baryonic_pk_zmax:
        raise ValueError(
            f"zmax={zmax} exceeds the baryonic emulator's valid range "
            f"(z <= {baryonic_pk_zmax}). Either lower zmax, or use "
            f"model_dmo from mcmc_model_no_baryons.py instead, which "
            f"isn't limited the same way."
        )
 
    mcmc_model_no_baryons.ZMIN = zmin
    mcmc_model_no_baryons.ZMAX = zmax
    mcmc_model_no_baryons.REDSHIFT_NODES = np.linspace(zmin, zmax, n_z_nodes)
    mcmc_model_no_baryons.Z_GRID = np.linspace(zmin, zmax, n_z_grid)
    mcmc_model_no_baryons.Z_GRID_R = np.linspace(0.0, zmax, n_z_grid_r)
    mcmc_model_no_baryons.k_custom = np.logspace(-3.5, np.log10(kmax), num=n_k)
 
 
```

```python
def build_covariance(zmin, zmax, kmax,
                      Omegam=0.31, w0=-1.0, s8=0.8, mc1=13.5,
                      Asurvey_deg2=ASURVEY_DEG2,
                      ngal_arcmin2=NGAL_ARCMIN2,
                      sigma_eps=SIGMA_EPS):
    """Compute the diagonal Gaussian covariance matrix for mcmc_model's
    ls_eval bins, at the given (zmin, zmax, kmax) and fiducial cosmology.
 
    Returns
    -------
    covmat : (Nbins, Nbins) ndarray, diagonal
    cl_fid : (Nbins,) ndarray, the fiducial C_l used to build it
    """
    _set_range(zmin, zmax, kmax)
 
    #cl_fid = mcmc_model_no_baryons.model_dmo(Omegam, w0, s8)
    cl_fid = mcmc_model.model(Omegam, w0, s8, mc1)

    
    # shape-noise power spectrum N_l (same for every l -- shot noise is
    # white/scale-independent)
    n_bar_sr = ngal_arcmin2 * ARCMIN2_PER_SR  # galaxies per steradian
    N_l = sigma_eps ** 2 / n_bar_sr
 
    # f_sky
    Asurvey_sr = Asurvey_deg2 * DEG2_TO_SR
    f_sky = Asurvey_sr / (4 * np.pi)
 
    # bin widths Delta_l, from mcmc_model's own ell-bin edges
    delta_l = np.array([hi - lo for lo, hi in mcmc_model.l_bins])
    ell = mcmc_model.ls_eval
 
    var_cl = 2.0 / ((2 * ell + 1) * f_sky * delta_l) * (cl_fid + N_l) ** 2
    covmat = np.diag(var_cl)
 
    print(f"Built covariance for zmin={zmin}, zmax={zmax}, kmax={kmax:.3f} h/Mpc")
    print(f"  f_sky = {f_sky:.4f}, N_l = {N_l:.4e}")
    print(f"  sigma(C_l)/C_l per bin: {np.sqrt(var_cl) / cl_fid}")
 
    return covmat, cl_fid
 
 
if __name__ == "__main__":
    # Example: reproduce your current (zmin=0.5, zmax=1.5) setup, with
    # kmax taken from mcmc_model's own current k_custom before we
    # overwrite it (grab it up front).
    default_kmax = mcmc_model.k_custom.max()
 
    covmat, cl_fid = build_covariance(zmin=0.5, zmax=1.5, kmax=default_kmax)
 
    out_name = f"covmat_zmin0.5_zmax1.5_kmax{default_kmax:.2f}.npy"
    np.save(out_name, covmat)
    print(f"Saved covariance matrix to {out_name}")
 
    # To generate a different one, just call build_covariance again with
    # different arguments, e.g.:
    #
    # covmat2, cl_fid2 = build_covariance(zmin=0.001, zmax=1.5, kmax=4.69)
    # np.save("covmat_zmin0.001_zmax1.5_kmax4.69.npy", covmat2)
 
```
